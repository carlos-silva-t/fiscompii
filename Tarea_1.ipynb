```python
def energia_cinetica(masa, velocidad_kmh):
    velocidad_ms = velocidad_kmh * 1000/3600
    energia_cinetica = 0.5 * masa * velocidad_ms**2
    return energia_cinetica

velocidad_kmh = 2
masa = 4

energia = energia_cinetica(masa, velocidad_kmh)
print(f"La energía cinética es : {energia} J")
```
