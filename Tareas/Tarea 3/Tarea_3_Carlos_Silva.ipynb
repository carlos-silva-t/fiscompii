```python
#Flujo base con ruido gaussiano con la desviacion estandar de 0.002.

Ruido = np.random.normal(loc= 0, scale= 0.002, size= len(t))
Flujo = 1.0 + Ruido
```

```python
#Simulacion del transito 1, disminuyendo el flujo en 0.02 entre los años 4.5 y 5.0.

Transito_1 = (t >= 4.5) & (t <= 5.0)
Flujo[Transito_1] -= 0.02
```

```python
#Simulacion del transito 2, disminuyendo el flujo en 0.015 entre los años 7.0 y 7.4.

Transito_2 = (t >= 7.0) & (t <= 7.4)
Flujo[Transito_2] -= 0.015
```

```python
#Def para generar la curva de luz.

def Curva_de_luz(Tiempo, Flujo):
    Flujo_medio = np.mean(Flujo)
    Sigma = np.std(Flujo)  
    Umbral = Flujo_medio - 3 * Sigma
```

```python
#Condición para detectar los puntos por debajo del umbral.

Puntos_bajos = Flujo < Umbral
plt.figure(figsize=(12, 6))
```

```python
#Puntos de la curva de luz

plt.plot(
    Tiempo[~Puntos_bajos],
    Flujo[~Puntos_bajos],
    "o",
    color="royalblue",
    markersize=4,
    label="Flujo observado",
    )
```

```python
#Puntos que están por debajo del umbral (transitos)

plt.plot(
    Tiempo[Puntos_bajos],
    Flujo[Puntos_bajos],
    "r^",
     markersize=6,
    label=r"Inmersión ($F < \bar{F} - 3\sigma$)",
    )
```

```python
#Línea horizontal del valor medio

plt.axhline(
    y=Flujo_medio,
    color="black",
    linestyle="--",
    linewidth=1.5,
    label=rf"Flujo medio ($\bar{{F}}$ = {Flujo_medio:.4f})",
    )
```

```python
# Línea horizontal del umbral.
plt.axhline(
    y=Umbral,
    color="red",
    linestyle=":",
    linewidth=1.5,
    label=rf"Umbral ($\bar{{F}} - 3\sigma$ = {Umbral:.4f})",
    )
```

```python
#Líneas verticales indicando el inicio y el fin de los tránsitos

plt.axvline(
    4.5,
    color="green",
    linestyle="-.",
    alpha=0.7,
    label="Inicio/Fin Tránsito 1",
    )
plt.axvline(5.0, color="green", linestyle="-.", alpha=0.7)
plt.axvline(
    7.0,
    color="purple",
linestyle="-.",
    alpha=0.7,
    label="Inicio/Fin Tránsito 2",
    )
plt.axvline(7.4, color="purple", linestyle="-.", alpha=0.7)
```

```python
# Configuración de etiquetas, título y leyenda

plt.xlabel("Tiempo [años]", fontsize=14)
plt.ylabel("Flujo normalizado", fontsize=14)
plt.title(
        "Curva de luz simulada con tránsitos", fontsize=16
    )
plt.legend(fontsize=14, loc="lower left")
plt.grid(True, linestyle=":", alpha=0.6)
plt.tight_layout()
plt.show()
```
